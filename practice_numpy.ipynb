# **Numpy Practices**

```python
import numpy as np
```

## Practice-1: array creation

```python
np.random.randint(1, 100, (3, 3))
```

## Practice -2: Array Filtering

```python
arr = np.array([5, 10, 15, 20, 25])

filtered_arr = arr[arr > 12]

print("filtered Array is", filtered_arr)
```

## Practice -3: Array Math operation

```python
gen_Arr = np.random.randint(1, 50, (1,6))

## operation performs
mean_Arr = np.mean(gen_Arr)
max_arr = np.max(gen_Arr)
min_arr = np.min(gen_Arr)
std_arr = np.std(gen_Arr)


# print of these values
print("Generated Array", gen_Arr)
print("Mean of this Array", mean_Arr)
print("Max value of this Array", max_arr)
print("Min Value of this Array", min_arr)
print("standard deviation of this Array", std_arr)

```

## Practice-4: Matrics multiplication

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

## operation
AB_multiply = np.dot(A, B)

## print

print("A =", A)
print("\n B =", B)
print("\n A * B =", AB_multiply)
```

## practice-5: reshaping

```python
ran_Arr = np.random.randint(1, 50, (1, 9))

reshaped_Arr = ran_Arr.reshape(3, 3)

print("Generated Array -> ", ran_Arr)
print("\n Reshaped Array -> ", reshaped_Arr)
```
